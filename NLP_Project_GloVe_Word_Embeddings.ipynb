```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# 1. Load the dictionary, exactly as shown in lecture:
glove_embeddings = {}
with open("glove.6B.100d.txt", 'r', encoding='utf-8') as f:
    for line in f:
        values = line.split()
        word = values[0]
        vector = list(map(float, values[1:]))
        glove_embeddings[word] = vector

my_words = ["king", "computer", "happy"]
```

```python
# 2.Use scikit-learn's cosine_similarity to find and print the 5 closest words
def most_similar(word, embeddings, topn=5):
    target = np.array(embeddings[word]).reshape(1, -1)
    scores = []
    for w, vec in embeddings.items():
        if w != word:
            sim = cosine_similarity(target, np.array(vec).reshape(1, -1))[0][0]
            scores.append((w, sim))
    scores.sort(key=lambda x: x[1], reverse=True)
    return scores[:topn]
```

```python
for w in my_words:
    print(f"Top 5 neighbors for '{w}':")
    print(most_similar(w, glove_embeddings))
    print("-" * 50)
```

1. Did the neighbors make sense to you? Give one example.

Yes, the neighbors made perfect sense.
For example, the top neighbors for the word "computer" include words like "software" and "pc", which are highly related technology terms.
This shows that the GloVe model successfully captures semantic relationships based on text co-occurrence.

Yes, the word "king" had some interesting neighbors.
While words like "prince" or "queen" are expected, it also brings up names of specific real-world kings or historical figures depending on the text corpus it was trained on.
This happens because the model learns from global news and text where the title "king" is heavily associated with specific famous individuals.
